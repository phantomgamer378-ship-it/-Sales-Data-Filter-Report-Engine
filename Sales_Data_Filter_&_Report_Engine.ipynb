```python
storetotal = random_sales.sum(axis =1)
storeaverage = random_sales.mean(axis = 1)
average = np.mean(storetotal)
below_avg = storetotal[storetotal<average]

print("Store total: ",storetotal)
print("Store average: ",storeaverage)



```

```python
month_total = random_sales.sum(axis=0)
s1_best = np.argsort(s1)[-3:]
print(s1_best)
print(s1)
print("stores 1 best months: ",months_n[s1_best])
s2_best = np.argsort(s2)[-3:]
s3_best = np.argsort(s3)[-3:]
print("store 2 best months: ",months_n[s2_best])
print("store 3 best months: ",months_n[s3_best])
```

```python

growth_targets = np.linspace(100000, 150000, 12)

print("MONTHLY SALES REPORT")
print("-" * 55)

for i in range(12):
    status = (
        "Target Met"
        if month_total[i] >= growth_targets[i]
        else "Target Missed"
    )

    print(
        f"{months_n[i]:12s} | "
        f"Sales: {month_total[i]:8d} | "
        f"Target: {growth_targets[i]:9.2f} | "
        f"{status}"
    )
```

```python

print("STORE-WISE ANNUAL REPORT")
print("-" * 55)
best_month_indices = np.argmax(random_sales, axis=1)
for i in range(10):
    print(
        f"Store {i + 1:2d} | "
        f"Annual Sales: {storetotal[i]:8d} | "
        f"Monthly Average: {storeaverage[i]:8.2f} | "
        f"Best Month: {months_n[best_month_indices[i]]}"
    )
```
